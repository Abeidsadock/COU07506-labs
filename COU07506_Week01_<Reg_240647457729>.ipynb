```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile
plt.rcParams["figure.figsize"] = (9, 3.5)
```

Task 1.1

```python
n = np.arange(-5, 11)
delta = (n == 0).astype(float)
# unit impulse
step  = (n >= 0).astype(float)
# unit step
ramp  = n * step
# unit ramp
expo  = 0.8**n * step
# real exponential
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, title in zip(axes, [delta, step, ramp, expo],
                        ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")
plt.tight_layout()
plt.show()
```

Task 1.2
Part 1: (a)ⁿ u(n) for a = 0.5, -0.8 and 1.1

```python
n = np.arange(0,20)

a_values = [0.5, -0.8, 1.1]

plt.figure(figsize=(12,4))

for i,a in enumerate(a_values):
    x = a**n

    plt.subplot(1,3,i+1)
    plt.stem(n,x)
    plt.title(f'a = {a}')
    plt.grid()

plt.tight_layout()
plt.show()
```

Task 1.2
Part 2: cos(2πn/8) and cos(2πn/7.5) for 0 ≤ n ≤ 40

```python
n = np.arange(0,50)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

plt.figure(figsize=(10,6))

plt.subplot(2,1,1)
plt.stem(n,x1)
plt.title('cos(2πn/8)')
plt.grid()

plt.subplot(2,1,2)
plt.stem(n,x2)
plt.title('cos(2πn/7.5)')
plt.grid()

plt.tight_layout()
plt.show()
```

Task 1.3

```python
import os
fname = "my_voice.wav"                 # change to your file name
if os.path.exists(fname):
    fs, x = wavfile.read(fname)
    if x.ndim > 1:                     # stereo -> mono
        x = x.mean(axis=1)
    x = x / np.max(np.abs(x))          # normalise to [-1, 1]
else:                                  # fallback test signal
    fs = 16000
    t = np.arange(0, 2, 1/fs)
    x = 0.5*np.sin(2*np.pi*440*t) * np.exp(-t)
t = np.arange(len(x)) / fs
print(f"fs = {fs} Hz, duration = {len(x)/fs:.2f} s, samples = {len(x)}")
plt.plot(t, x); plt.xlabel("time (s)"); plt.ylabel("amplitude")
plt.title("My recording"); plt.show()
```
